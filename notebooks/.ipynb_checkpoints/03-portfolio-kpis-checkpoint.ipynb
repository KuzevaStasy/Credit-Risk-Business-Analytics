# 03 — Portfolio KPIs
**Credit Risk Business Analytics**

Key performance indicators for the loan portfolio — default rates, income distribution, credit amounts, debt ratios, and demographic breakdown.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# ── Paths ──────────────────────────────────────────────
ROOT        = os.path.abspath("..")
CLEAN_PATH  = os.path.join(ROOT, "data", "processed", "cleaned_data.csv")
FIGURES_DIR = os.path.join(ROOT, "reports", "figures")
os.makedirs(FIGURES_DIR, exist_ok=True)

# ── Style ──────────────────────────────────────────────
plt.rcParams.update({
    "figure.figsize"   : (12, 5),
    "axes.spines.top"  : False,
    "axes.spines.right": False,
    "axes.grid"        : True,
    "grid.alpha"       : 0.3,
    "font.size"        : 11,
})
BLUE  = "#2563eb"
RED   = "#ef4444"
LIGHT = "#93c5fd"

print("Setup complete")
```

```python
# ── Load cleaned data ──────────────────────────────────
df = pd.read_csv(CLEAN_PATH)
print(f"Loaded: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
print(f"Default rate: {df['TARGET'].mean():.2%}")
```

## 1 — Default Rate Overview

```python
counts = df["TARGET"].value_counts()
labels = ["Non-Default (0)", "Default (1)"]
colors = [BLUE, RED]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Pie chart
axes[0].pie(
    counts.values, labels=labels, colors=colors,
    autopct="%1.1f%%", startangle=90,
    wedgeprops={"edgecolor": "white", "linewidth": 2}
)
axes[0].set_title("Portfolio Composition", fontweight="bold")

# Bar chart
bars = axes[1].bar(labels, counts.values, color=colors)
axes[1].bar_label(bars, fmt="{:,.0f}", padding=5)
axes[1].set_title("Default vs Non-Default Count", fontweight="bold")
axes[1].set_ylabel("Number of Loans")
axes[1].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x:,.0f}"))

plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, "default_rate.png"), dpi=150)
plt.show()
print("Saved: default_rate.png")
```

## 2 — Default Rate by Risk Tier

```python
if "RISK_TIER" in df.columns:
    tier_stats = df.groupby("RISK_TIER", observed=True).agg(
        Count       = ("TARGET", "count"),
        Default_Rate= ("TARGET", "mean")
    ).reset_index()

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Default rate by tier
    tier_colors = [RED, "#f97316", "#facc15", "#22c55e"]
    bars = axes[0].bar(tier_stats["RISK_TIER"], tier_stats["Default_Rate"] * 100,
                       color=tier_colors)
    axes[0].bar_label(bars, fmt="%.1f%%", padding=4)
    axes[0].set_title("Default Rate by Risk Tier", fontweight="bold")
    axes[0].set_ylabel("Default Rate (%)")

    # Count by tier
    bars2 = axes[1].bar(tier_stats["RISK_TIER"], tier_stats["Count"],
                        color=tier_colors)
    axes[1].bar_label(bars2, fmt="{:,.0f}", padding=4)
    axes[1].set_title("Loan Count by Risk Tier", fontweight="bold")
    axes[1].set_ylabel("Number of Loans")
    axes[1].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x:,.0f}"))

    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_DIR, "default_by_risk_tier.png"), dpi=150)
    plt.show()
    print("Saved: default_by_risk_tier.png")
```

## 3 — Income Distribution

```python
if "AMT_INCOME_TOTAL" in df.columns:
    # Cap at 99th percentile to remove extreme outliers
    cap = df["AMT_INCOME_TOTAL"].quantile(0.99)
    income_df = df[df["AMT_INCOME_TOTAL"] <= cap]

    fig, ax = plt.subplots()
    ax.hist(income_df[income_df["TARGET"] == 0]["AMT_INCOME_TOTAL"],
            bins=50, alpha=0.6, color=BLUE, label="Non-Default")
    ax.hist(income_df[income_df["TARGET"] == 1]["AMT_INCOME_TOTAL"],
            bins=50, alpha=0.6, color=RED,  label="Default")
    ax.set_title("Income Distribution by Default Status", fontweight="bold")
    ax.set_xlabel("Annual Income")
    ax.set_ylabel("Count")
    ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x/1000:.0f}K"))
    ax.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_DIR, "income_distribution.png"), dpi=150)
    plt.show()
    print("Saved: income_distribution.png")
```

## 4 — Credit Amount Distribution

```python
if "AMT_CREDIT" in df.columns:
    cap = df["AMT_CREDIT"].quantile(0.99)
    credit_df = df[df["AMT_CREDIT"] <= cap]

    fig, ax = plt.subplots()
    ax.hist(credit_df[credit_df["TARGET"] == 0]["AMT_CREDIT"],
            bins=50, alpha=0.6, color=BLUE, label="Non-Default")
    ax.hist(credit_df[credit_df["TARGET"] == 1]["AMT_CREDIT"],
            bins=50, alpha=0.6, color=RED,  label="Default")
    ax.set_title("Credit Amount Distribution by Default Status", fontweight="bold")
    ax.set_xlabel("Credit Amount")
    ax.set_ylabel("Count")
    ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x/1000:.0f}K"))
    ax.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_DIR, "credit_distribution.png"), dpi=150)
    plt.show()
    print("Saved: credit_distribution.png")
```

## 5 — Age Distribution

```python
if "AGE_YEARS" in df.columns:
    fig, ax = plt.subplots()
    ax.hist(df[df["TARGET"] == 0]["AGE_YEARS"],
            bins=40, alpha=0.6, color=BLUE, label="Non-Default")
    ax.hist(df[df["TARGET"] == 1]["AGE_YEARS"],
            bins=40, alpha=0.6, color=RED,  label="Default")
    ax.set_title("Age Distribution by Default Status", fontweight="bold")
    ax.set_xlabel("Age (years)")
    ax.set_ylabel("Count")
    ax.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_DIR, "age_distribution.png"), dpi=150)
    plt.show()
    print("Saved: age_distribution.png")
```

## 6 — Debt-to-Income by Default Status

```python
if "DEBT_TO_INCOME" in df.columns:
    cap = df["DEBT_TO_INCOME"].quantile(0.99)
    dti_df = df[df["DEBT_TO_INCOME"] <= cap]

    dti_by_target = dti_df.groupby("TARGET")["DEBT_TO_INCOME"].median()

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Histogram
    axes[0].hist(dti_df[dti_df["TARGET"] == 0]["DEBT_TO_INCOME"],
                 bins=50, alpha=0.6, color=BLUE, label="Non-Default")
    axes[0].hist(dti_df[dti_df["TARGET"] == 1]["DEBT_TO_INCOME"],
                 bins=50, alpha=0.6, color=RED,  label="Default")
    axes[0].set_title("Debt-to-Income Distribution", fontweight="bold")
    axes[0].set_xlabel("Debt-to-Income Ratio")
    axes[0].set_ylabel("Count")
    axes[0].legend()

    # Median comparison
    bars = axes[1].bar(["Non-Default", "Default"], dti_by_target.values,
                       color=[BLUE, RED])
    axes[1].bar_label(bars, fmt="%.2f", padding=4)
    axes[1].set_title("Median Debt-to-Income by Status", fontweight="bold")
    axes[1].set_ylabel("Median DTI Ratio")

    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_DIR, "debt_to_income.png"), dpi=150)
    plt.show()
    print("Saved: debt_to_income.png")
```

## 📝 Portfolio KPIs Summary

| KPI | Finding |
|---|---|
| Overall default rate | ~8% of the portfolio |
| Highest risk tier | High Risk — default rate 3–4× the average |
| Income | Defaulters tend to have lower median income |
| Age | Younger borrowers show higher default rates |
| Debt-to-Income | Defaulters carry higher DTI ratios |
