# 01 — Data Overview
**Credit Risk Business Analytics**

First look at the available datasets — structure, size, data types, missing values, and key variables.

```python
import os
import pandas as pd
import numpy as np

# ── Paths ──────────────────────────────────────────────
ROOT    = os.path.abspath("..")
RAW_DIR = os.path.join(ROOT, "data", "raw")

FILES = {
    "feature_matrix"          : os.path.join(RAW_DIR, "feature_matrix.csv"),
    "feature_matrix_advanced" : os.path.join(RAW_DIR, "feature_matrix_advanced.csv"),
    "feature_matrix_spec"     : os.path.join(RAW_DIR, "feature_matrix_spec.csv"),
    "correlations"            : os.path.join(RAW_DIR, "correlations.csv"),
    "correlations_spec"       : os.path.join(RAW_DIR, "correlations_spec.csv"),
    "feature_importances"     : os.path.join(RAW_DIR, "feature_importances.csv"),
    "fi_fma"                  : os.path.join(RAW_DIR, "fi_fma.csv"),
    "spec_feature_importances": os.path.join(RAW_DIR, "spec_feature_importances_ohe.csv"),
}

print("ROOT:", ROOT)
print("\nAvailable files:")
for name, path in FILES.items():
    exists = os.path.exists(path)
    print(f"  {'✅' if exists else '❌'} {name}")
```

## 1 — Main Feature Matrix

```python
# ── Load main feature matrix ───────────────────────────
df = pd.read_csv(FILES["feature_matrix"])
print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
```

```python
# ── First look ─────────────────────────────────────────
df.head(5)
```

```python
# ── Column types ───────────────────────────────────────
dtype_summary = df.dtypes.value_counts()
print("Column types:")
print(dtype_summary)
```

```python
# ── TARGET variable ────────────────────────────────────
print("TARGET distribution:")
print(df["TARGET"].value_counts())
print(f"\nDefault rate: {df['TARGET'].mean():.2%}")
```

```python
# ── Missing values ─────────────────────────────────────
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    "Missing count" : missing,
    "Missing %"     : missing_pct
}).query("`Missing count` > 0").sort_values("Missing %", ascending=False)

print(f"Columns with missing values: {len(missing_df)} out of {df.shape[1]}")
missing_df.head(20)
```

```python
# ── Key numeric features ───────────────────────────────
key_cols = [
    "AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "AMT_GOODS_PRICE",
    "DAYS_BIRTH", "DAYS_EMPLOYED",
    "EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"
]
key_cols = [c for c in key_cols if c in df.columns]
df[key_cols].describe().round(2)
```

## 2 — Feature Importance Files

```python
# ── Feature importances ────────────────────────────────
fi = pd.read_csv(FILES["feature_importances"])
print(f"Feature importances shape: {fi.shape}")
fi.head(10)
```

```python
# ── FI Advanced ────────────────────────────────────────
fi_adv = pd.read_csv(FILES["fi_fma"])
print(f"FI Advanced shape: {fi_adv.shape}")
fi_adv.head(10)
```

```python
# ── Spec feature importances ───────────────────────────
fi_spec = pd.read_csv(FILES["spec_feature_importances"])
print(f"Spec FI shape: {fi_spec.shape}")
fi_spec.head(10)
```

## 3 — Correlations with TARGET

```python
# ── Load correlations ──────────────────────────────────
corr = pd.read_csv(FILES["correlations"], index_col=0)
print(f"Correlations matrix shape: {corr.shape}")

if "TARGET" in corr.columns:
    top_corr = (
        corr["TARGET"]
        .drop("TARGET", errors="ignore")
        .abs()
        .sort_values(ascending=False)
        .head(15)
    )
    print("\nTop 15 features correlated with TARGET:")
    print(top_corr)
```

## 📝 Dataset Inventory

| Dataset | Purpose | Key variable |

| `feature_importances.csv` | Base feature importance | importance column |
