# 02 — Data Cleaning
**Credit Risk Business Analytics**

Preparing the feature matrix for analysis — handling missing values, fixing types, engineering key business metrics, and saving a clean analytical dataset.

```python
import os
import pandas as pd
import numpy as np

# ── Paths ──────────────────────────────────────────────
ROOT       = os.path.abspath("..")
RAW_PATH   = os.path.join(ROOT, "data", "raw",       "feature_matrix.csv")
CLEAN_PATH = os.path.join(ROOT, "data", "processed", "cleaned_data.csv")

os.makedirs(os.path.dirname(CLEAN_PATH), exist_ok=True)
print("Clean file will be saved to:", CLEAN_PATH)
```

```python
# ── Load raw data ──────────────────────────────────────
df = pd.read_csv(RAW_PATH)
print(f"Raw shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
```

## 1 — Remove Invalid Records

```python
# ── Drop rows with no TARGET ───────────────────────────
before = len(df)
df = df.dropna(subset=["TARGET"])
print(f"Removed {before - len(df):,} rows with missing TARGET")
print(f"Remaining: {len(df):,} rows")
```

```python
# ── TARGET as integer ──────────────────────────────────
df["TARGET"] = df["TARGET"].astype(int)
print("TARGET distribution:")
print(df["TARGET"].value_counts())
print(f"\nDefault rate: {df['TARGET'].mean():.2%}")
```

## 2 — Fix Days Columns

```python
# ── DAYS_BIRTH → Age in years (positive) ──────────────
if "DAYS_BIRTH" in df.columns:
    df["AGE_YEARS"] = (-df["DAYS_BIRTH"] / 365).round(1)
    print(f"Age range: {df['AGE_YEARS'].min()} – {df['AGE_YEARS'].max()} years")

# ── DAYS_EMPLOYED → fix anomalous value (365243 = unemployed flag) ──
if "DAYS_EMPLOYED" in df.columns:
    anomaly_mask = df["DAYS_EMPLOYED"] == 365243
    print(f"\nDAYS_EMPLOYED anomalies (365243): {anomaly_mask.sum():,} rows")
    df["DAYS_EMPLOYED"] = df["DAYS_EMPLOYED"].replace(365243, np.nan)
    df["YEARS_EMPLOYED"] = (-df["DAYS_EMPLOYED"] / 365).round(1)
    df["IS_UNEMPLOYED"]  = anomaly_mask.astype(int)
    print(f"Years employed range: {df['YEARS_EMPLOYED'].min()} – {df['YEARS_EMPLOYED'].max()}")
```

## 3 — Business KPI Columns

```python
# ── Debt-to-Income ratio ───────────────────────────────
if all(c in df.columns for c in ["AMT_CREDIT", "AMT_INCOME_TOTAL"]):
    df["DEBT_TO_INCOME"] = (df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]).round(3)
    print(f"DEBT_TO_INCOME — median: {df['DEBT_TO_INCOME'].median():.2f}")

# ── Annuity-to-Income ratio ────────────────────────────
if all(c in df.columns for c in ["AMT_ANNUITY", "AMT_INCOME_TOTAL"]):
    df["ANNUITY_TO_INCOME"] = (df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]).round(3)
    print(f"ANNUITY_TO_INCOME — median: {df['ANNUITY_TO_INCOME'].median():.2f}")

# ── Credit-to-Goods ratio ──────────────────────────────
if all(c in df.columns for c in ["AMT_CREDIT", "AMT_GOODS_PRICE"]):
    df["CREDIT_TO_GOODS"] = (df["AMT_CREDIT"] / df["AMT_GOODS_PRICE"].replace(0, np.nan)).round(3)
    print(f"CREDIT_TO_GOODS — median: {df['CREDIT_TO_GOODS'].median():.2f}")
```

## 4 — Handle Missing Values

```python
# ── Drop columns with >60% missing ────────────────────
threshold = 0.6
missing_pct = df.isnull().mean()
cols_to_drop = missing_pct[missing_pct > threshold].index.tolist()

print(f"Columns with >{threshold*100:.0f}% missing: {len(cols_to_drop)}")
df = df.drop(columns=cols_to_drop)
print(f"Shape after dropping: {df.shape[0]:,} × {df.shape[1]:,}")
```

```python
# ── Impute remaining numeric columns with median ───────
numeric_cols = df.select_dtypes(include=[np.number]).columns
remaining_missing = df[numeric_cols].isnull().sum()
cols_to_impute = remaining_missing[remaining_missing > 0].index.tolist()

for col in cols_to_impute:
    df[col] = df[col].fillna(df[col].median())

print(f"Imputed {len(cols_to_impute)} numeric columns with median")
print(f"Remaining missing values: {df.isnull().sum().sum()}")
```

## 5 — Risk Tier Column

```python
# ── EXT_SOURCE combined score → Risk Tier ─────────────
ext_cols = [c for c in ["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"] if c in df.columns]

if ext_cols:
    df["EXT_SOURCE_MEAN"] = df[ext_cols].mean(axis=1).round(4)

    df["RISK_TIER"] = pd.cut(
        df["EXT_SOURCE_MEAN"],
        bins=[0, 0.35, 0.50, 0.65, 1.0],
        labels=["High Risk", "Medium-High", "Medium-Low", "Low Risk"],
        include_lowest=True
    )

    print("Risk Tier distribution:")
    print(df["RISK_TIER"].value_counts().sort_index())

    print("\nDefault rate by Risk Tier:")
    print(df.groupby("RISK_TIER", observed=True)["TARGET"].mean().apply(lambda x: f"{x:.2%}"))
```

## 6 — Final Check & Save

```python
# ── Final shape ────────────────────────────────────────
print(f"Final shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
print(f"Missing values remaining: {df.isnull().sum().sum()}")
df.head(3)
```

```python
# ── Save cleaned data ──────────────────────────────────
# Sample to max 15,000 rows for dashboard performance
df_sample = df.sample(n=min(15000, len(df)), random_state=42)
df_sample.to_csv(CLEAN_PATH, index=False)

print(f"Saved {len(df_sample):,} rows to: {CLEAN_PATH}")
```

## 📝 Cleaning Summary

| Step | Action |
|---|---|
| 1 | Removed rows with missing TARGET |
| 2 | Converted DAYS_BIRTH → AGE_YEARS |
| 3 | Fixed DAYS_EMPLOYED anomaly (365243 → unemployed flag) |
| 4 | Created DEBT_TO_INCOME, ANNUITY_TO_INCOME, CREDIT_TO_GOODS |
| 5 | Dropped columns with >60% missing |
| 6 | Imputed remaining numeric columns with median |
| 7 | Created EXT_SOURCE_MEAN and RISK_TIER segmentation |
